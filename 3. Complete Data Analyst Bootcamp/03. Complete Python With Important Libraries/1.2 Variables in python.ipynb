# Variables In Python

Variables are fundamental elements in programming used to store data that can be referenced and manipulated throughout a program. In Python, variables are created when uou assign a value to them, and they do not need explicit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable 

```python
a = 100
print(a)
```

Declaration and assigning variables

```python
# Declaration and assigning variables

age = 22
height = 5.9
name = "Pankaj"
is_student = True

# Printing the variables

print("Age:", age)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)
```

# Naming Conventions

```python
# Variable name should be descriptive and should not start with a number or special character. It can contain letters, numbers, and underscores.    
# They must start with a letter or an underscore. They cannot start with a number or special character.
# variable names are case-sensitive, meaning that 'age' and 'Age' would be considered different variables.

# Valid variable names:
my_variable = 10
_variable = 20
Age = 30


```

```python
# Invalid variable names:

# 1variable = 10  # Starts with a number
# my-variable = 20  # Contains a hyphen
# my variable = 30  # Contains a space
```

# Variable Types

Python is dynamically typed, type of a variable is determined at runtime

```python
age = 22
height = 5.9
name = "Pankaj"
is_student = True

print(type(age))        # Output: <class 'int'>
print(type(height))     # Output: <class 'float'>
print(type(name))       # Output: <class 'str'>
```

# Type Conversion

```python
age = 22
print(type(age))

age_str = str(age)
print(type(age_str))
```

```python
age = '22'
print(type(age))
print(type(int(age)))
print(age)
print(type(age))
```

```python
height = 5.9
h = int(height)

print(h)
```

# Dynamic Typing

Python allows the type of a variable to change as the program executes

```python
var = 10
print(var, type(var))

var = 'Hello'
print(var, type(var))

var = 3.14
print(var, type(var))
```

# input() function

input() function : Use to take inputs from the user and by default it takes input as string str

```python
# Input
age = input("What is your Age : ")
print(age, type(age))
age = int(input("What is your Age : "))
print(age, type(age))
```
