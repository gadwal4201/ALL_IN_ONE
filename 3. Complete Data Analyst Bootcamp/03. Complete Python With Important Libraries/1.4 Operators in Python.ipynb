# Deep Dive into Operators 

1. Arithmetic Operators

```python


num1 = 10
num2 = 5

sum = num1 + num2 # addition
diff = num1 - num2 # Subtraction
product = num1 * num2 # Multiplication
quotient = num1 / num2 # Division
floor_div = num1 // num2 # Floor Division provides only int
modulus = num1 % num2 # Modulus/ remainder
exponent = num1 ** num2 # Exponent a**b means a to the power b 

print(f"Sum of {num1} and {num2} is : {sum}")
print(f"Difference of {num1} and {num2} is : {diff}")
print(f"Multiplication {num1} and {num2} is : {product}")
print(f"Division of {num1} and {num2} is : {quotient}")
print(f"Floor Division of {num1} and {num2} is : {floor_div}")
print(f"Modulus of {num1} and {num2} is : {modulus}")
print(f"Exponent of {num1} and {num2} is : {exponent}")
```

2. Comparison Operators

```python
a = 10 # Assignment Operator =
b = 10
print(a == b) # Equal to Operator ==
print(a != b) # Not Equal to !=
print(a > b) # Greater than >
print(a < b) # Less than <
print(a >= b) # Greater than equal to
print(a <= b) # Less than equal to

```

3. Logical Operators

```python
# and, or, not

x = True
y = False

print(x and y)
print(x or y)
print(not x)
print(not y)
```
