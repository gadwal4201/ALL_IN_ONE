# Tuples in Python

Tuples are ordered collection of items that are immutable. They are similar to lists, but their immutability makes them different

```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1,2,3,4,5,6])
print(numbers)
print(type(numbers))
```

```python
list_from_tuple = list(numbers)
print(list_from_tuple)
print(type(list_from_tuple))
```

```python
mixed_tuple = (1, 2, 3, "Hello", 4.5)
print(mixed_tuple)
```

Immutable Nature of tuple

```python
# Tuples are immutable, so we cannot change their values. However, we can create a new tuple by concatenating two tuples or repeating a tuple multiple times.

lst = [1, 2, 3, 4, 5]
print(lst)

lst[1] = "Pankaj"
print(lst)
```

```python
numbers[1] = "Pankaj" # This will raise an error because tuples are immutable and we cannot change their values.
print(numbers)
```

# Tuple Methods

```python
numbers = (100,15,23,86,400,23,825)
```

```python
print(numbers.count(23)) # Count the number of occurrences of 1 in the tuple
```

```python
print(numbers.index(23)) # Find the index of the first occurrence of 3 in the tuple
```

# Packing & Unpacking Tuples

```python
packed_tuple = 1, 2, 3, "Hello", 4.5
print(packed_tuple)
```

```python
a, b, c, d, e = packed_tuple
print(a)
print(b)
print(c)
print(d)
```

# Unpacking with Star(*)

```python
numbers = (1, 2, 3, 4, 5, 6)
first, *middle = numbers
print(first)
print(middle)
# print(last)
```

```python
numbers = (1, 2, 3, 4, 5, 6)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

# Nested Tuple

```python
lst = [1,2,3,["hello",500],5,[True,False,[1,2,["Python","Java",1500]]]]
print(lst)
```

```python
tpl = tuple(lst)
print(tpl)
```

```python
lst[3][1]
```

```python
tpl[3][1]
```

```python
lst[-1][-1][-1][0]
```

```python
tpl[-1][-1][-1][0]
```

# Iterating over nested tuples

```python
for sub_tpl in tpl:
    for i in sub_tpl:
        print(i,end=" ")
    print()
```
