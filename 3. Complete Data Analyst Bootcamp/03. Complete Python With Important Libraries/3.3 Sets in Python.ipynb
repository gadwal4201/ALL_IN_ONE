# Sets in Python

Sets are a built-in data type in Python used to store unordered collections of unique elements. They are useful for performing mathematical set operations like union, intersection, and difference. Sets are defined using curly braces {} or the set() constructor.

```python
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(my_empty_set)
print(type(my_empty_set))
```

```python
lst_to_set = set([1, 2, 3, 4, 5, 5, 5]) # TypeCasting a list to a set removes duplicates
print(lst_to_set)
```

# Mathematical Operation

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(set1)
print(set2)
```

1. Union Operation

```python
set1.union(set2) # Returns a new set with all unique elements from both sets
```

2. Intersection Operation

```python
set1.intersection(set2) # Returns a new set with elements common to both sets
```

3. Intersection_Update

```python
set1.intersection_update(set2) # Updates set1 with elements common to both sets

print(set1)
```

4. Difference Operation

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(set1)
print(set2)
```

```python
print(set1.difference(set2))
print(set2.difference(set1))
```

```python
set1.difference_update(set2)
```

```python
print(set1)
```

5. Symmetric Difference

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(set1)
print(set2)
```

```python
set1.symmetric_difference(set2) # removes common elements and shows only unique elements
```

```python
set1.symmetric_difference_update(set2)
print(set1)
```
