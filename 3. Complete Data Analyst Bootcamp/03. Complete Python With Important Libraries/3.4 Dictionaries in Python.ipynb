# Dictionaries in Python

Dictionaries are ordered collections of items. They store data in a key value pairs. Keys must be unique and immutable(e.g. strings, numbers, or tuples), while values can be of any type.

```python
# creating dictionaries

empty_dict = {}
print(type(empty_dict))
print(empty_dict)
```

```python
# creating dictionaries

empty_dict = dict()
print(type(empty_dict))
print(empty_dict)
```

```python
students = {'name':'Pankaj','age':32,'grade':'A'}
print(students)
```

```python
Repeat_students = {'name':'Pankaj','name':'Chirag'}
print(Repeat_students) # shows last updated name
```

```python
Repeat_students['name']
```

# Accessing using get() method

```python
students.get('grade')
```

```python
print(students.get('Last_name')) # by default gives NOne
```

```python
print(students.get('last_name',"Not Available"))
```

# Modifying Dictionaries elements

```python
# Dictionaries are mutable, so you can add, update or delete elements

students['age'] = 22  # Update values in dict
print(students)
```

```python
students['address'] = "India"  # adding new key value pair
print(students)
```

```python
del students['grade'] # delete key and value pair
print(students)
```

# Dictionaries Methods

```python
keys = students.keys()
print(keys)
values = students.values()
print(values)
items = students.items()
print(items)
```

# shallow copy

```python
students_copy = students
print(students)
print(students_copy)
```

```python
students['name'] = 'Chirag'
print(students)
print(students_copy)
```

```python
new_copy_students = students.copy() # use copy() to make a seperate copy
print(students)
print(new_copy_students)
```

```python
new_copy_students['name'] = 'Nikhil'
print(new_copy_students)
print(students)
```

```python
# Iterating over dictionaries

for i in students.keys():
    print(i)
```

```python
for i in students.values():
    print(i)
```

```python
# Iterate over key value pair

for key,value in students.items():
    print(f"{key}:{value}")
```

# Nested Dictionaries

```python
std = {
    'student1':{'name':'Pankaj','age':22},
    'student2':{'name':'chirag','age':38},
}
```

```python
print(std)
```

```python
print(std['student1']['name'])
print(std['student2']['age'])
```

```python
# Iterating over nested dict

for student_id,student_info in std.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f'{key}:{value}')
```

# Dictionaries Comprehensions

```python
squares = {x:x**2 for x in range(10,15) if x%2 == 0}
print(squares)
```

```python
# Use a dictionary to count the frequency of elements in list

numbers = [1,2,3,4,5,6,6,4,7,8,9,1,2,3,4,5,7,8]

frequency = {}
for i in numbers:
    if i in frequency:
        frequency[i]+=1
    else:
        frequency[i]=1
print(frequency)
```
