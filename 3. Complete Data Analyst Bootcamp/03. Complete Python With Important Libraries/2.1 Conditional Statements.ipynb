# Conditional Statements

1. if Statement

```python
age = 22

if age >= 18:
    print("You are an adult and can vote.")
```

2. else Statement

```python
if age < 18:
    print("You are a minor and cannot vote.")
else:
    print("You are an adult and can vote.")
```

3. elif Statement

```python
# elif stands for "else if" and is used to check multiple conditions. It allows you to test additional conditions if the previous ones were not met.

if age < 13:
    print("You are a child.")
elif age < 18:
    print("You are a teenager.")
elif age < 65:
    print("You are an adult.")
else:
    print("You are a senior citizen.")
```

4. Nested Conditional Statements

```python
# Nested if statements are if statements placed inside another if statement. They allow you to check for additional conditions within a specific block of code.

num = int(input("Enter a number: "))

if num >= 0:
    print("The number is non-negative.")
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
else:
    print("The number is negative.")
```

```python
# Practical Example : Determine if a year is a leap year or not using conditional statements.

year = int(input("Enter a year: "))

if year >= 0:
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                print(f"{year} is a leap year.")
            else:
                print(f"{year} is not a leap year.")
        else:
            print(f"{year} is a leap year.")
    else:
        print(f"{year} is not a leap year.")
else:
    print("Please enter a valid year.")
```

# Task

```python
n1 = int(input("Enter 1st number : "))
n2 = int(input("Enter 2nd number : "))
op = input("Enter Operation (+,-,*,/,%,//) : ")

if op == "+":
    print(f"Addition of {n1} and {n2} is : {n1+n2}")
elif op == "-":
    print(f"Subtraction of {n1} and {n2} is : {n1-n2}")
elif op == "*":
    print(f"Multiplication of {n1} and {n2} is : {n1*n2}")
elif op == "/":
    if n2 != 0:
        print(f"Division of {n1} and {n2} is : {n1/n2}")
    else:
        print("Error: Division by zero is not allowed.")
elif op == "%":
    if n2 != 0:
        print(f"Modulus of {n1} and {n2} is : {n1%n2}")
    else:
        print("Error: Modulus by zero is not allowed.")
elif op == "//":
    if n2 != 0:
        print(f"Floor Division of {n1} and {n2} is : {n1//n2}")
    else:
        print("Error: Floor Division by zero is not allowed.")
```
