# Loops in Python

Loops in python are used to execute a block of code repeatedly until a certain condition is met. There are two main types of loops in Python: the `for` loop and the `while` loop.



# range() Function

```python
# range() function is used to generate a sequence of numbers. It can take one, two, or three arguments.

# range(start, stop, step) generates numbers from start to stop (exclusive) with a step size of step. If only one argument is provided, it is treated as the stop value, and the start defaults to 0. The step defaults to 1 if not specified.

for i in range(5):  # Generates numbers from 0 to 4
    print(i)
```

# for Loop

```python
# for loops are used to iterate over a sequence (like a list, tuple, dictionary, set, or string). The syntax for a for loop is:
# for variable in sequence:
#     # Code to be executed

for i in range(1,10,2):  # Generates numbers from 1 to 9 with a step of 2
    print(i)
```

```python
for i in range(10,0,-2):  # Generates numbers from 10 to 2 with a step of -2
    print(i)
```

```python
str = "Hello, World!"
for char in str:  # Iterates over each character in the string
    print(char)
```

# while Loop

```python
# While loop is used to execute a block of code repeatedly as long as a given condition is true. The syntax for a while loop is:
# while condition:
#     # Code to be executed

count = 0
while count < 5:  # Executes the loop as long as count is less than 5
    print(count)
    count += 1  # Increment count by 1
```

```python
count = 0

while count % 2 == 0:  # Executes the loop as long as count is even
    print(count)
    count += 1  # Increment count by 1
```

# Loop Control Statements


1. break
2. continue
3. pass

```python
# break statement is used to exit a loop prematurely when a certain condition is met. It can be used in both for and while loops.

for i in range(10):
    if i == 5:  # If i equals 5, exit the loop
        break
    print(i)
```

```python
# continue statement is used to skip the current iteration of a loop and move on to the next iteration. It can be used in both for and while loops.
for i in range(10):
    if i == 5:  # If i equals 5, skip the rest of the loop body
        continue
    print(i)
```

```python
for i in range(10):
    if i % 2 ==0:  # If i is even, skip the rest of the loop body
        continue
    print(i)
```

```python
# pass statement is a null operation; it is used when a statement is syntactically required but you do not want any command or code to execute. It can be used in both for and while loops.
for i in range(10):
    if i == 5:  # If i equals 5, do nothing and continue to the next iteration
        pass
    print(i)
```

```python
for i in range(10):
    if i == 5:  # If i equals 5, do nothing and continue to the next iteration
        pass
    else:
        print(i)
```

# Nested Loops

```python
# Nested loops are loops within loops. The inner loop is executed for each iteration of the outer loop.

for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```

```python
# Example : Calculate the sum of first N natural numbers using a for loop and a while loop.

N = int(input("Enter a positive integer: "))
# Using a for loop

sum_for = 0
for i in range(1, N + 1):
    sum_for += i
    
# Using a while loop
sum_while = 0
count = 1
while count <= N:
    sum_while += count
    count += 1
    
print(f"Sum of first {N} natural numbers using for loop: {sum_for}")
print(f"Sum of first {N} natural numbers using while loop: {sum_while}")
```

```python
# Example : Prime numbers b/w 1 and 100

for num in range(1,100 + 1):
    if num > 1:
        for i in range(2,num):
            if num % i == 0:
                break
        else:
            print(num)
```
